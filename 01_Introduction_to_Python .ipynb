## Introduction to Python 🐍


## 1. **What is Python..?**
- Python is high-level interpreted language and object-oriented programming language.
- It is simple, readable, and easy-to-understand, and it is widely used.


---

## 2. Who Created Python?
- Python was created by **Guido van Rossum**.
- He started working on Python in the late 1980s.

---

## 3. When Was Python Created?
- Python development started in **1989**.
- The first public version was released in **1991**.

---

## 4. First Version of Python

- The first public version of Python was **Python 0.9.0**.
- It was released in **February 1991**.

---

## 5. Python 1.0

- Python 1.0 was released in **1994**.
- It introduced many important features and helped Python become more useful.

---

## 6. Python 2.0
- Python 2.0 was released in **2000**.
- It introduced several improvements and new features to the language.

---

## 7. Python 3.0
- Python 3.0 was released in **2008**.
- It was a major update designed to improve Python and make the language cleaner.

---

## 8. Python Versions
- Python 0.9.0 → 1991
- Python 1.0 → 1994
- Python 2.0 → 2000
- Python 3.0 → 2008
