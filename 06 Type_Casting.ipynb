## 1. What is Type Casting?

- Type Casting means converting the value of one data type into another data type.
- For example: converting an `int` to a `float`, or a `str` to an `int`.

```python
a = 10        # a is an integer
b = 3.5       # b is a float

print(type(a))   # shows the type of a -> <class 'int'>
print(type(b))   # shows the type of b -> <class 'float'>
```

## 2. Implicit Type Casting (Automatic)
- Python automatically converts one data type into another
- when needed during an operation.

```python
                            # Implicit casting example
x = 5              # x is an integer
y = 2.5            # y is a float

z = x + y          # when we add int + float, Python automatically converts int to float

print(z)          # Output: 7.5
print(type(z))    # Output: <class 'float'>  (automatic conversion happened here)
```

## 3. Explicit Type Casting (Manual)

Here the programmer manually converts the type using built-in functions.
Common functions: `int()`, `float()`, `str()`

```python
                        # Converting float to int manually
num1 = 9.8                     # num1 is a float value
num2 = int(num1)                # int() function manually converts it to int

print(num2)                      # Output: 9  (the decimal part .8 is cut off)
print(type(num2))                # Output: <class 'int'>
```

```python
# Converting string to int
age_str = "25"             # age_str is a string
age_int = int(age_str)     # converted the string to an int

print(age_int + 5)         # Output: 30  (now we can do math with it)
print(type(age_int))       # Output: <class 'int'>
```

```python
# Converting int to string
marks = 90                 # marks is an integer
marks_str = str(marks)     # str() converts it into a string

print("My marks are " + marks_str)   # now we can join it with other text
print(type(marks_str))               # Output: <class 'str'>
```

## 4. Limitations of Type Casting

1. **Data Loss** - converting float to int removes the decimal part.
2. **Invalid Conversion Error** - if a string doesn't contain a valid number, you get an error.
3. **Precision Loss** - converting very large or very precise numbers can lose accuracy.

```python
# Limitation 1: Data Loss example
value = 9.99                # original float value
converted = int(value)      # int() only keeps the whole number part, no rounding

print(converted)            # Output: 9  (the .99 part is lost, not rounded)
```

```python
# Limitation 2: Invalid Conversion Error example
text = "hello"        # this string is not a number

# Uncommenting the line below will cause an ERROR:
# num = int(text)     # ValueError: invalid literal for int() with base 10: 'hello'

print("Only a numeric string can be converted using int()")
```
