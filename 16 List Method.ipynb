# Python List Methods


---

### append()
- Adds one item to the end of a list.

````python
numbers = [10, 20, 30]
numbers.append(40)   # Adds 40 at the end
print(numbers)
````

````python
sales = [100, 200]
sales.append(300)    # Adds new sales record at the end
print(sales)
````

### extend()

- Adds all items from another list (one by one) to the end of a list.

````python
numbers = [10, 20, 30]
numbers.extend([40, 50])   # Adds 40 and 50 as separate items
print(numbers)
````

````python
fruits = ["apple", "banana"]
more_fruits = ["mango", "orange"]
fruits.extend(more_fruits)   # Adds all items from more_fruits
print(fruits)
````

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

list1.append([4, 5])   # Adds [4, 5] as one single item
list2.extend([4, 5])   # Adds 4 and 5 as separate items

print(list1)   # [1, 2, 3, [4, 5]]
print(list2)   # [1, 2, 3, 4, 5]
````

### insert()
- Adds an item at a specific position (index) in the list.

````python
numbers = [10, 20, 30]
numbers.insert(1, 15)   # Inserts 15 at index 1
print(numbers)
````

````python
names = ["Amit", "Karan"]
names.insert(0, "Zoya")   # Inserts Zoya at the beginning
print(names)
````

### remove()
- Removes the first matching item (by value) from the list.

````python
numbers = [10, 20, 30, 20]
numbers.remove(20)   # Removes the first 20 found
print(numbers)
````

````python
fruits = ["apple", "banana", "cherry"]
fruits.remove("banana")   # Removes banana from the list
print(fruits)
````

### pop()

- Removes and returns an item from a list.
- By default, it removes the last item.
- We can also give an index to remove a specific item.

````python
numbers = [10, 20, 30, 40]

numbers.pop(1)   # Removes the item at index 1
print(numbers)

numbers.pop()    # Removes the last item
print(numbers)
````

````python
numbers = [10, 20, 30]
removed_item = numbers.pop()   # Removes the last item
print(removed_item)
print(numbers)
````

````python
numbers = [10, 20, 30]
removed_item = numbers.pop(1)   # Removes item at index 1
print(removed_item)
print(numbers)
````

````python
numbers = [10, 20, 30]

numbers.remove(20)   # Removes value 20
print(numbers)

numbers2 = [10, 20, 30]
numbers2.pop(1)       # Removes item at index 1
print(numbers2)
````

### clear()

- Removes all items from the list, making it empty.

````python
numbers = [10, 20, 30]
numbers.clear()   # Empties the list
print(numbers)
````

````python
data = ["a", "b", "c"]
data.clear()   # Removes all items from data
print(data)
````

### index()

- Returns the position (index) of the first matching item in the list.

````python
numbers = [10, 20, 30, 40]
position = numbers.index(30)   # Finds position of 30
print(position)
````

````python
fruits = ["apple", "banana", "cherry"]
position = fruits.index("cherry")   # Finds position of cherry
print(position)
````

### count()

- Counts how many times a value appears in the list.

````python
numbers = [10, 20, 20, 30, 20]
total = numbers.count(20)   # Counts how many times 20 appears
print(total)
````

````python
marks = [90, 85, 90, 70, 90]
total = marks.count(90)   # Counts how many times 90 appears
print(total)
````

### sort()


- Arranges list items in ascending order by default.
- reverse=True arranges items in descending order.

````python
numbers = [40, 10, 30, 20]

numbers.sort()             # Sorts the list in ascending order
print(numbers)

numbers.sort(reverse=True) # Sorts the list in descending order
print(numbers)
````

````python
```python
numbers = [40, 10, 30, 20]

numbers.sort(reverse=True)   # Sorts the list in descending order
print(numbers)
````

````python
numbers = [40, 10, 30, 20]
numbers.sort()   # Sorts the list in ascending order
print(numbers)
````

````python
names = ["Zoya", "Amit", "Karan"]
names.sort()   # Sorts names alphabetically
print(names)
````

### reverse()

- Reverses the order of items in the list. It changes the original list.

````python
numbers = [10, 20, 30, 40]
numbers.reverse()   # Reverses the order of items
print(numbers)
````

````python
fruits = ["apple", "banana", "cherry"]
fruits.reverse()   # Reverses the list of fruits
print(fruits)
````

````python
numbers = [30, 10, 40, 20]

numbers.sort()      # Arranges numbers from smallest to largest
print(numbers)       # [10, 20, 30, 40]

numbers.reverse()    # Flips the current order
print(numbers)       # [40, 30, 20, 10]
````

### copy()

- Creates a copy of the list. Changes made to the copy do not affect the original list.

````python
original = [10, 20, 30]
new_list = original.copy()   # Makes a separate copy
new_list.append(40)          # Changing the copy
print(original)   # Original list is unchanged
print(new_list)   # Copy has the new item
````

````python
fruits = ["apple", "banana"]
fruits_copy = fruits.copy()   # Makes a copy of fruits
fruits_copy.append("mango")
print(fruits)         # Original stays the same
print(fruits_copy)    # Copy has mango added
````
