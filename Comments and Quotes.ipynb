## 1. What is a Comment? (Comment kya hota hai?)

- A **comment** is a line of text in your code that Python **does not run**. It is only there for humans to read.

### **Why do we use comments?**
- To explain what the code is doing
- To leave notes for yourself or other people
- To temporarily "turn off" a line of code without deleting it

### **Types of comments in Python:**
1. **Single-line comment** — starts with `#`
2. **Multi-line comment** — written using `''' '''` or `""" """` (technically it's an unused string, but it works like a comment)

```python
## This is a single-line comment
print("Hello World")   # you can also write a comment after code, like this

## The line below is a comment, so it will NOT run
## print("This will not print")
```

'''
This is a multi-line comment.
You can write as many lines as you want here.
None of this will run.
'''

print("Code after the comment still runs")


## 2. Reassigning a Value (Variable ki value badalna)

In Python, you can give a variable a **new value** at any time. This is called **reassigning**.
The new value simply **replaces** the old one.

Python is beginner-friendly here: a variable can hold a number first, and later you can put a text (string) in the same variable.

```python
x = 10
print("Step 1:", x)

x = 25          # reassigning: old value 10 is replaced by 25
print("Step 2 (after reassigning):", x)

x = "Hello"     # now x holds text instead of a number
print("Step 3 (reassigned again):", x)
```

```python
# You can also reassign using the variable's own old value
count = 5
count = count + 1   # short way to write this: count += 1
print("New count:", count)
```

## 3. Single Quotes, Double Quotes and Triple Quotes

In Python, we use **quotes** to create text, called a **string**.

| Quote Type | Symbol | Example |
|---|---|---|
| Single quote | `' '` | `'Hello'` |
| Double quote | `" "` | `"Hello"` |
| Triple quote | `''' '''` or `""" """` | `'''Hello'''` |

**Single quotes vs Double quotes:**
- They both do the **exact same job** — they create a string
- There is no real difference between them
- Choosing one over the other becomes useful when your text already has a quote inside it

**Triple quotes:**
- Used to write a string across **multiple lines**
- Used for **comments** (as we saw above)
- Used for **docstrings** — a short description written just below a function to explain what it does

```python
# Single quote
name1 = 'Rahul'

# Double quote
name2 = "Rahul"

print(name1, name2)
print(name1 == name2)   # both are exactly the same, so this prints True
```

```python
# If your text has a single quote inside it, use double quotes outside
sentence1 = "It's a sunny day"
print(sentence1)

# If your text has a double quote inside it, use single quotes outside
sentence2 = 'She said "Hello"'
print(sentence2)
```

```python
# Triple quotes -> string that spans multiple lines
para = '''This is line one
This is line two
This is line three'''
print(para)
```

```python
def greet():
    """This is a docstring. It explains what the function does."""
    print("Hello!")

greet()
print(greet.__doc__)   # this shows the docstring
```
