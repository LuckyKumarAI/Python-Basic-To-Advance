# Python String Methods


- A method is a predefined function that belongs to an object or data type and is used to perform a specific task.
- A method can be used with the same or compatible data type.


## `upper()`

- `upper()` converts all letters in a string to uppercase (capital letters).

```python
# This code converts the text to uppercase letters
text = "Hello World"

result = text.upper()

print(result)
```

```python
name = "lucky kumar"
print(name.upper())           # All the letters in the text are now capital letters.
```

## `lower()`


- `lower()` converts all letters in a string to lowercase (small letters).

```python
# This code converts the text to lowercase letters
text = "Hello World"

result = text.lower()

print(result)
```

```python
city = "INDIA"
print(city.lower())                   # All the letters in the text are now small letters.
```

## `capitalize()`


- `capitalize()` makes the first character of the string uppercase and changes the rest of the letters to lowercase.

```python
# This code capitalizes only the first letter
text = "hello python world"

result = text.capitalize()

print(result)
```

```python
text = "india is my country"

print(text.capitalize())                    # Only the first letter of the whole sentence becomes capital.
                                            # The rest of the letters become lowercase.
```

## `title()`

- `title()` makes the first letter of each word uppercase.

```python
# This code capitalizes the first letter of every word
text = "hello python world"

result = text.title()

print(result)
```

```python
text = "india is my country"

print(text.title())                  # The first letter of every word is now capital.
```

## `swapcase()`

- `swapcase()` changes uppercase letters to lowercase and lowercase letters to uppercase.

```python
# This code swaps the case of every letter
text = "Hello World"

result = text.swapcase()

print(result)
```

```python
text = "iNdiA iS yY cOuNTrY"
                                    # Capital letters become small, and small letters become capital.
print(text.swapcase())                  
```

## `strip()`

- `strip()` removes extra spaces from both the left side and the right side of a string.

```python
# This code removes spaces from both sides
text = "   Hello World   "

result = text.strip()

print(result)
```

```python
name = "   Lucky   "

print(name.strip())                  # Removes extra spaces from the beginning and end.
```

## `lstrip()`

- `lstrip()` removes extra spaces only from the left side of a string.

```python
# This code removes spaces from the left side only
text = "   Hello World   "

result = text.lstrip()

print(result)
```

```python
text = "   Hello World"

print(text.lstrip())    # Removes extra spaces from the beginning of the string.
```

```python
text = "###Hello World"

print(text.lstrip("#"))    # Removes # from the beginning of the string.
```

## `rstrip()`

- `rstrip()` removes extra spaces only from the right side of a string.

```python
# This code removes spaces from the right side only
text = "   Hello World   "

result = text.rstrip()

print(result)
```

```python
text = "Hello World   "

print(text.rstrip())    # Removes extra spaces from the end of the string.
```

## `replace()`

- `replace()` changes a part of the string with another word or text.

```python
text = "I like Python"

print(text.replace("Python", "SQL"))    # Replaces Python with SQL.
```

```python
text = "Hello World"

print(text.replace("World", "Lucky"))    # Replaces World with Lucky.
```

### `replace()` Method

The `replace()` method is used to replace one part of a string with another.

**Syntax:**
`string.replace(old, new, count)`

- **old** → The text that we want to replace.
- **new** → The new text that we want to add.
- **count** → The maximum number of times the replacement should happen.

**Example:**

`text.replace("apple", "mango", 2)`

Here:
- `"apple"` → old text
- `"mango"` → new text
- `2` → replace only the first 2 occurrences

## `split()`

- `split()` breaks a string into a list of smaller pieces (words), using a space as the default separator.

```python
# This code breaks the sentence into a list of words
text = "Hello World Python"

result = text.split()

print(result)
```

```python
text = "I love Python"

print(text.split())    # Splits the string into a list using spaces.
```

```python
data = "Python|SQL|Excel|Power BI"

print(data.split("|"))    # Splits the string using | as the delimiter.
```

## `find()`

- `find()` searches for a word or letter inside a string and gives its starting position. If the word is not found, it gives `-1`.

```python
# This code searches for the word "World" inside the text
text = "Hello World"

result = text.find("World")

print(result)
```

```python
text = "I love Python"

print(text.find("Python"))    # Returns the starting index of Python.
```

```python
text = "I love Python"

print(text.find("Java"))    # Returns -1 if the text is not found.
```

```python
# This code searches for a word that does not exist in the text
text = "Hello World"                         
                                 # Since "Python" is not present in "Hello World", `find()` gives `-1`.
print(text.find("Python"))
```

## `index()`

- `index()` also searches for a word or letter inside a string and gives its starting position. But if the word is not found, it gives an error.

```python
# This code searches for the word "World" inside the text
text = "Hello World"

result = text.index("World")

print(result)
```

```python
# This code searches for a word that does not exist in the text
text = "Hello World"

print(text.index("Python"))
```

**Explanation:**
This example gives an error because `"Python"` is not present in the text. `index()` does not return `-1` like `find()`. Instead, it stops the program with a `ValueError`.

## `count()`

- `count()` tells how many times a word or letter appears inside a string.

```python
# This code counts how many times "o" appears in the text
text = "Hello World"

result = text.count("o")

print(result)
```

```python
text = "I love Python and I love SQL"

print(text.count("love"))    # Counts how many times "love" appears in the string.
```

```python
text = "banana"

print(text.count("a"))    # Counts how many times "a" appears in the string.
```

## `startswith()`

- `startswith()` checks if a string begins with a certain word or letter. It gives `True` or `False`.

```python
# This code checks if the text starts with "Hello"
text = "Hello World"
                                        # The text does start with "Hello", so the answer is `True`.
result = text.startswith("Hello")

print(result)
```

```python
text = "Python is easy"

print(text.startswith("Python"))    # Checks if the string starts with "Python".
```

```python
text = "Python is easy"

print(text.startswith("Java"))    # Returns False if the string does not start with "Java".
```

## `endswith()`

- `endswith()` checks if a string ends with a certain word or letter. It gives `True` or `False`.

```python
# This code checks if the text ends with "World"
text = "Hello World"

result = text.endswith("World")

print(result)
```

```python
file_name = "data.csv"

print(file_name.endswith(".csv"))    # Checks if the file name ends with ".csv".
```

```python
text = "I love Python"

print(text.endswith("Python"))    # Checks if the string ends with "Python".
```

## `isalpha()`

- `isalpha()` checks if all the characters in a string are letters. It gives `True` or `False`.

```python
# This code checks if the text has only letters
text = "Hello"

result = text.isalpha()

print(result)
```

```python
text = "Python123"

print(text.isalpha())    # Returns False because the string contains numbers.
```

## `isdigit()`

- `isdigit()` checks if all the characters in a string are digits (numbers). It gives `True` or `False`.

```python
# This code checks if the text has only digits
text = "12345"

result = text.isdigit()
                            # "12345" has only digits, so the answer is `True`.
print(result)
```

```python
text = "123abc"

print(text.isdigit())    # Returns False because the string contains letters.
```

```python
text = "12345"

print(text.isdigit())    # Checks if all characters are digits.
```

## `isalnum()`

- `isalnum()` checks if all the characters in a string are letters or numbers (no spaces or symbols). It gives `True` or `False`.

```python
# This code checks if the text has only letters and numbers
text = "Hello123"

result = text.isalnum()

print(result)
```

```python
text = "Python123"

print(text.isalnum())    # Checks if all characters are letters or numbers.
```

```python
text = "Python 123"

print(text.isalnum())    # Returns False because the string contains a space.
```

## `islower()`

- `islower()` checks if all the letters in a string are lowercase (small letters). It gives `True` or `False`.

```python
# This code checks if the text is fully in lowercase
text = "hello world"

result = text.islower()

print(result)
```

```python
text = "hello"

print(text.islower())    # Checks if the letters are in lowercase.
```

```python
text = "Hello"

print(text.islower())    # Returns False because H is uppercase.
```

## `isupper()`

- `isupper()` checks if all the letters in a string are uppercase (capital letters). It gives `True` or `False`.

```python
# This code checks if the text is fully in uppercase
text = "HELLO WORLD"

result = text.isupper()

print(result)
```

```python
text = "HELLO"

print(text.isupper())    # Checks if the letters are in uppercase.
```

```python
text = "Hello"

print(text.isupper())    # Returns False because some letters are lowercase.
```

## `istitle()`

- `istitle()` checks if the first letter of every word in a string is capital, and the rest of the letters are small. It gives `True` or `False`.

```python
# This code checks if the text follows title case
text = "Hello World"

result = text.istitle()

print(result)
```

```python
text = "Hello World"

print(text.istitle())    # Checks if each word starts with a capital letter.
```

```python
text = "hello world"

print(text.istitle())    # Returns False because the words do not start with capital letters.
```

## `isspace()`

- `isspace()` checks if a string has only spaces. It gives `True` or `False`.

```python
# This code checks if the text has only spaces
text = "   "

result = text.isspace()

print(result)
```

**Explanation:**
The text has only spaces, so the answer is `True`.
