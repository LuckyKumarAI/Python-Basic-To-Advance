# Python List Functions, Indexing, and Slicing


## List Functions


### 1. `len()`
Returns the number of items in a list.

```python
fruits = ["apple", "banana", "cherry"]

print(len(fruits))   # Counts total items in the list -> 3
```

### 2. `max()`
Returns the largest value in a list.

```python
numbers = [10, 25, 3, 47, 8]

print(max(numbers))   # Finds the biggest number -> 47
```

### 3. `min()`
Returns the smallest value in a list.

```python
numbers = [10, 25, 3, 47, 8]

print(min(numbers))   # Finds the smallest number -> 3
```

### 4. `sum()`
Adds up all the values in a list.

```python
numbers = [10, 20, 30]

print(sum(numbers))   # Adds all numbers -> 60
```

### 5. `sorted()`
Returns a **new sorted list**. The original list stays the same.

```python
numbers = [40, 10, 30, 20]

print(sorted(numbers))        # Sorts in ascending order -> [10, 20, 30, 40]
print(numbers)                 # Original list is unchanged
```

```python
names = ["Zoya", "Amit", "Karan"]

print(sorted(names))          # Sorts alphabetically
```

### 6. `list()`
Creates a list. It can convert other data types (like a string or range) into a list.

```python
letters = list("abc")   # Converts a string into a list of characters
print(letters)           # ['a', 'b', 'c']
```

```python
numbers = list(range(5))   # Converts a range into a list
print(numbers)              # [0, 1, 2, 3, 4]
```

### 7. `any()`
Returns `True` if **at least one** item in the list is `True` (or non-zero).

```python
values = [0, 0, 5, 0]

print(any(values))   # True, because 5 is non-zero
```

### 8. `all()`
Returns `True` only if **all** items in the list are `True` (or non-zero).

```python
values = [1, 2, 3, 4]

print(all(values))   # True, because none of the values is 0
```

```python
values = [1, 0, 3]

print(all(values))   # False, because one value is 0
```

## List Indexing

Indexing means accessing a single item from a list using its position number.
- Positive index starts from `0` (first item).
- Negative index starts from `-1` (last item).


```python
numbers = [10, 20, 30, 40, 50]

print(numbers[0])      # First element -> 10
print(numbers[2])      # Third element -> 30
print(numbers[-1])     # Last element -> 50
print(numbers[-2])     # Second last element -> 40
```

---

## List Slicing

Slicing means getting a **part** of a list (a sub-list) instead of just one item.

Basic form: `list[start:end:step]`
- `start` -> where to begin (included)
- `end` -> where to stop (not included)
- `step` -> how many items to skip


```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])     # Gets elements from index 1 to 3 -> [20, 30, 40]
print(numbers[:3])      # Gets first 3 elements -> [10, 20, 30]
print(numbers[2:])      # Gets elements from index 2 to the end -> [30, 40, 50]
print(numbers[:])       # Gets the whole list -> [10, 20, 30, 40, 50]
```

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[::2])     # Gets every second element -> [10, 30, 50]
print(numbers[::-1])    # Reverses the list -> [50, 40, 30, 20, 10]
```
