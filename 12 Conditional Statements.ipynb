## 1. What are Conditional Statements?

A **conditional statement** is a way of telling the computer:

> "If something is true, do this. Otherwise, do something else."

In real life, we make decisions all the time:
- If it is raining, take an umbrella.
- If age is greater than or equal to 18, you are eligible to vote.
- If marks are greater than or equal to 40, you pass the exam.

Python uses conditional statements to make these kinds of decisions in code.

**Why do we use conditional statements?**
- Because a program should not always do the same thing. It should behave differently depending on the situation (the data).


````python
# Example 1: Check if a person is eligible to vote
age = 20

if age >= 18:
    print("You are eligible to vote")
````

````python
# Example 2: Check if a student has passed
marks = 45

if marks >= 40:
    print("You passed the exam")
````

⚠️ Note: `==` is used to **compare** two values. `=` is used to **assign** a value. These are different!


````python
# Comparison operators in action

a = 10
b = 5

print(a > b)    # Is 10 greater than 5? -> True
print(a < b)    # Is 10 less than 5? -> False
print(a >= 10)  # Is 10 greater than or equal to 10? -> True
print(a <= 5)   # Is 10 less than or equal to 5? -> False
print(a == 10)  # Is 10 equal to 10? -> True
print(a != b)   # Is 10 not equal to 5? -> True
````

````python
# Another simple example with names (comparing text)

name1 = "Amit"
name2 = "Amit"

print(name1 == name2)  # Are both names the same? -> True
print(name1 != name2)  # Are the names different? -> False
````

## 3. `if` Statement

The `if` statement checks a condition.
If the condition is `True`, the code inside the `if` block runs.
If the condition is `False`, the code inside is simply skipped.

**Basic Syntax:**
```python
if condition:
    # code to run if condition is True
```

Notice:
- The line must end with a colon `:`
- The next line must be **indented** (a few spaces before the code)

````python
# Example 1: Check if a number is positive
number = 7

if number > 0:
    print("The number is positive")
````

````python
# Example 2: Check if it is raining
is_raining = True

if is_raining:
    print("Take an umbrella")
````

## 4. `if-else` Statement

Sometimes we want the program to do one thing when the condition is `True`,
and a **different** thing when the condition is `False`.

This is where `if-else` is used.

**Difference between `if` and `if-else`:**
- `if` alone only runs code when the condition is `True`. If it is `False`, nothing happens.
- `if-else` gives a second option, so something always happens, either the `if` part or the `else` part.

**Basic Syntax:**
```python
if condition:
    # runs when condition is True
else:
    # runs when condition is False
```


````python
# Example 1: Check if a number is even or odd
number = 9

if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
````

````python
# Example 2: Check if a student has passed or failed
marks = 30

if marks >= 40:
    print("You passed")
else:
    print("You failed")
````

## 5. `if-elif-else` Statement

Sometimes there are **more than two** possible outcomes.
For this, we use `elif` (short for "else if").

**What is `elif`?**
It lets us check another condition if the first `if` condition was `False`.

**Why do we use `elif`?**
Because real-life situations often have more than 2 options. For example, grades can be A, B, C, or D — not just pass/fail.

**How Python checks conditions:**
Python checks conditions **from top to bottom**, one by one.
As soon as one condition is `True`, Python runs that block and **skips the rest**.

**Basic Syntax:**
```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False and condition2 is True
else:
    # runs if none of the above are True
```


````python
# Example 1: Marks to Grade
marks = 75

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 40:
    print("Grade: C")
else:
    print("Grade: F")
````

````python
# Example 2: Age to Category
age = 15

if age < 13:
    print("Category: Child")
elif age < 20:
    print("Category: Teenager")
else:
    print("Category: Adult")
````

## 6. Multiple `elif` Conditions

We are not limited to just one `elif`.
We can use **as many `elif` statements as we need** to check multiple conditions.

Python still checks them from top to bottom, and stops at the first `True` condition.



````python
# Example 1: Weather temperature description
temperature = 5

if temperature >= 35:
    print("It is very hot")
elif temperature >= 25:
    print("It is warm")
elif temperature >= 15:
    print("It is cool")
elif temperature >= 0:
    print("It is cold")
else:
    print("It is freezing")
````

````python
# Example 2: Day number to Day name (simple version)
day_number = 3

if day_number == 1:
    print("Monday")
elif day_number == 2:
    print("Tuesday")
elif day_number == 3:
    print("Wednesday")
elif day_number == 4:
    print("Thursday")
else:
    print("Some other day")
````

## 7. Nested `if`

A **nested `if`** means writing an `if` statement **inside another `if` statement**.

We use this when we want to check a second condition, but only **after** the first condition is already `True`.

**Basic Syntax:**
```python
if condition1:
    if condition2:
        # runs only when both condition1 and condition2 are True
```


````python
# Example 1: Check eligibility for a driving license
age = 20
has_id_proof = True

if age >= 18:
    if has_id_proof:
        print("You can apply for a driving license")
````

````python
# Example 2: Check login details
username_correct = True
password_correct = True

if username_correct:
    if password_correct:
        print("Login successful")
````

## 8. Logical Operators with Conditional Statements

Logical operators let us **combine multiple conditions** together.

### `and`
`and` means **both** conditions must be `True` for the result to be `True`.

### `or`
`or` means **at least one** condition must be `True` for the result to be `True`.

### `not`
`not` simply **reverses** the result. `True` becomes `False`, and `False` becomes `True`.



````python
# and - Example 1: Check if eligible for a job (age AND qualification)
age = 25
has_degree = True

if age >= 18 and has_degree:
    print("Eligible for the job")
````

````python
# and - Example 2: Check if a number is between 10 and 20
number = 15

if number > 10 and number < 20:
    print("The number is between 10 and 20")
````

````python
# or - Example 1: Check if today is a weekend
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It is a weekend")
````

````python
# or - Example 2: Check if a person can get a discount
is_student = False
is_senior_citizen = True

if is_student or is_senior_citizen:
    print("You get a discount")
````

````python
# not - Example 1: Check if it is NOT raining
is_raining = False

if not is_raining:
    print("You can go outside without an umbrella")
````

````python
# not - Example 2: Check if a user is NOT logged in
is_logged_in = False

if not is_logged_in:
    print("Please log in first")
````

## 9. Combining Comparison and Logical Operators

We can use **comparison operators** (like `>`, `<`, `==`) together with
**logical operators** (like `and`, `or`) to check more detailed conditions.


````python
# Example 1: Check if a person can watch a movie (age check with and)
age = 16
has_parent_permission = True

if age >= 18 or (age >= 13 and has_parent_permission):
    print("You can watch the movie")
else:
    print("You cannot watch the movie")
````

````python
# Example 2: Check for a valid exam score entry
score = 85

if score >= 0 and score <= 100:
    print("Valid score")
else:
    print("Invalid score")
````

## 10. `in` and `not in` with Conditions

`in` checks if a value **exists** inside something, like a word inside a sentence.
`not in` checks if a value **does not exist** inside something.

These are very useful with conditional statements.


````python
# Example 1: Check if a letter is in a word
word = "python"

if "y" in word:
    print("The letter 'y' is in the word")
````

````python
# Example 2: Check if a fruit is not in a sentence
sentence = "I like apples and bananas"

if "mango" not in sentence:
    print("Mango is not mentioned in the sentence")
````

## 11. Truthy and Falsy Values

In Python, values can act like `True` or `False` even if they are not written as `True`/`False`.

- Values like `0`, `""` (empty text), and `None` are treated as **Falsy** (they behave like `False`).
- Almost all other values, like `10` or `"hello"`, are treated as **Truthy** (they behave like `True`).

This means we can write `if 10:` instead of `if 10 != 0:`.



````python
# Example 1: A non-zero number is Truthy
number = 10

if number:
    print("This runs because 10 is Truthy")
````

````python
# Example 2: Zero is Falsy
number = 0

if number:
    print("This will NOT print")
else:
    print("This runs because 0 is Falsy")
````

## 12. Conditional Expression / Ternary Operator

This is a **short, one-line way** of writing a simple `if-else` statement.

**Syntax:**
```python
value_if_true if condition else value_if_false
```

**How is it different from normal `if-else`?**
- Normal `if-else` takes multiple lines and is used to run different blocks of code.
- The ternary expression is used to quickly choose **one value** in a single line, and is often used with `print()` or when storing a value in a variable.



````python
# Example 1: Check if a number is even or odd (in one line)
number = 7

result = "Even" if number % 2 == 0 else "Odd"
print(result)
````

````python
# Example 2: Check pass or fail (in one line)
marks = 55

result = "Pass" if marks >= 40 else "Fail"
print(result)
````

## 13. Common Beginner Mistakes

Here are some common mistakes beginners make while writing conditional statements.

### Mistake 1: Using `=` instead of `==`
`=` is for assigning a value. `==` is for comparing values.

```python
# Wrong
if age = 18:
    print("Eligible")
```

```python
# Correct
if age == 18:
    print("Eligible")
```

### Mistake 2: Forgetting the colon `:`
Every `if`, `elif`, and `else` line must end with a colon.

```python
# Wrong
if age >= 18
    print("Eligible")
```

```python
# Correct
if age >= 18:
    print("Eligible")
```

### Mistake 3: Incorrect Indentation
Python uses indentation (spaces) to know which code belongs inside the `if` block.

```python
# Wrong
if age >= 18:
print("Eligible")
```

```python
# Correct
if age >= 18:
    print("Eligible")
```

### Mistake 4: Wrong Condition (logic mistake)
Sometimes the condition itself is incorrect for what we actually want to check.

```python
# Wrong (this checks for FAIL, but the message says PASS)
if marks < 40:
    print("You passed")
```

```python
# Correct
if marks >= 40:
    print("You passed")
```

### Mistake 5: Writing `if` without proper indentation for multiple lines
All lines inside the same block must have the **same** indentation.

```python
# Wrong (second line has different indentation)
if age >= 18:
    print("Eligible")
      print("Welcome")
```

```python
# Correct
if age >= 18:
    print("Eligible")
    print("Welcome")
```

### Mistake 6: Confusing `and` and `or`
- Use `and` when **both** conditions must be `True`.
- Use `or` when **at least one** condition must be `True`.

```python
# Wrong logic: this will almost never be True for a normal age
if age < 0 and age > 150:
    print("Invalid age")
```

```python
# Correct logic: use 'or' here instead
if age < 0 or age > 150:
    print("Invalid age")
```


## 14. Difference Between Conditional Statements

| Statement       | Use                                                |
|------------------|-----------------------------------------------------|
| `if`             | When we want to check one condition                 |
| `if-else`        | When there are two possible outcomes                |
| `if-elif-else`   | When there are multiple conditions                  |
| Nested `if`      | When one condition is checked inside another         |

**`if` vs `if-else`:**
`if` only runs code when the condition is `True`, and does nothing otherwise.
`if-else` always runs something — either the `if` part or the `else` part.

**`if-else` vs `if-elif-else`:**
`if-else` only handles **two** outcomes.
`if-elif-else` can handle **many** outcomes by checking multiple conditions one after another.

**`if-elif-else` vs Nested `if`:**
`if-elif-else` checks conditions **one after another**, and only one block runs.
Nested `if` checks a condition **inside** another condition — both conditions must be checked together, one after the other, not as alternatives.
