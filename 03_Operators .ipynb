# Python Operators
A beginner-friendly notebook covering Arithmetic, Comparison, Logical, and Assignment operators in Python, with a comment explaining every operator.

## 1. Arithmetic Operators
Used to perform mathematical calculations.

| Operator | Name | Example |
|---|---|---|
| `+` | Addition | `a + b` |
| `-` | Subtraction | `a - b` |
| `*` | Multiplication | `a * b` |
| `/` | Division (gives float result) | `a / b` |
| `//` | Floor Division (gives whole number, drops decimal part) | `a // b` |
| `%` | Modulus (gives the remainder after division) | `a % b` |
| `**` | Exponent (raises a number to a power) | `a ** b` |

```python
a = 10
b = 3
```

```python
print("Addition:", a + b)          # + adds two numbers together
```

```python
print("Subtraction:", a - b)       # - subtracts second number from first
```

```python
print("Multiplication:", a * b)    # * multiplies two numbers
```

```python
print("Division:", a / b)          # / divides and always gives a float (decimal) answer
```

```python
print("Floor Division:", a // b)   # // divides and rounds DOWN to the nearest whole number (no decimal)
```

```python
print("Modulus:", a % b)           # % gives the remainder left over after division
```

```python
print("Exponent:", a ** b)         # ** raises the first number to the power of the second (10 x 10 x 10)
```

## 2. Comparison Operators
Used to compare two values. The result is always a Boolean (`True` or `False`).

| Operator | Name | Example |
|---|---|---|
| `==` | Equal to | `a == b` |
| `!=` | Not equal to | `a != b` |
| `>` | Greater than | `a > b` |
| `<` | Less than | `a < b` |
| `>=` | Greater than or equal to | `a >= b` |
| `<=` | Less than or equal to | `a <= b` |

```python
print(a == b)   # == checks if both values are EQUAL
```

```python
print(a != b)   # != checks if both values are NOT equal
```

```python
print(a > b)    # > checks if left value is GREATER than right value
```

```python
print(a < b)    # < checks if left value is SMALLER than right value
```

```python
print(a >= b)   # >= checks if left value is GREATER THAN OR EQUAL to right value
```

```python
print(a <= b)   # <= checks if left value is SMALLER THAN OR EQUAL to right value
```

---

## 3. Logical Operators
Used to combine two or more conditions together.

| Operator | Name | Meaning |
|---|---|---|
| `and` | Logical AND | True only if BOTH conditions are True |
| `or` | Logical OR | True if AT LEAST ONE condition is True |
| `not` | Logical NOT | Reverses the result (True becomes False, and vice versa) |

```python
x = True
y = False
```

```python
print(x and y)   # and -> result is True ONLY IF BOTH x and y are True
```

```python
print(x or y)    # or -> result is True IF AT LEAST ONE of x or y is True
```

```python
print(not x)     # not -> flips the value, so True becomes False
```

## 4. Assignment Operators
Used to assign values to variables, often combined with arithmetic to update a variable's value.

| Operator | Name | Same As |
|---|---|---|
| `=` | Assign | `c = 5` |
| `+=` | Add and assign | `c = c + 3` |
| `-=` | Subtract and assign | `c = c - 2` |
| `*=` | Multiply and assign | `c = c * 4` |
| `/=` | Divide and assign | `c = c / 3` |
| `//=` | Floor divide and assign | `c = c // 3` |
| `%=` | Modulus and assign | `c = c % 3` |
| `**=` | Exponent and assign | `c = c ** 2` |

```python
c = 5
print(c)   # = simply stores the value 5 in c
```

```python
c += 3   # += means c = c + 3 (adds 3 to c and updates it)
print(c)
```

```python
c -= 2   # -= means c = c - 2 (subtracts 2 from c and updates it)
print(c)
```

```python
c *= 4   # *= means c = c * 4 (multiplies c by 4 and updates it)
print(c)
```

```python
c /= 3   # /= means c = c / 3 (divides c by 3 and updates it, gives float)
print(c)
```

```python
c //= 3   # //= means c = c // 3 (floor divides c by 3 and updates it)
print(c)
```

```python
c %= 2    # %= means c = c % 2 (finds remainder of c divided by 2, updates it)
print(c)
```

```python
c **= 2   # **= means c = c ** 2 (raises c to the power of 2, updates it)
print(c)
```
