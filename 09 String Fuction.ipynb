# String Functions


## Section 1: What is a Function?

- A function is a reusable block of code that performs a specific task and can work with different types of data.

**Why functions are useful:**
- They save us from writing the same code again and again.
- They make code easier to read and understand.
- Python already gives us many ready-made functions called **built-in functions**.



````python
# print() is a built-in function
# It shows (displays) the message on the screen

print("Hello, Python learner!")

# Here, "Hello, Python learner!" is the input given to the function
# The function's job is to display it on the screen
````

## Section 2: Built-in String-Related Functions

1. `str()` - Converts a value into a string
2. `len()` - Finds the number of characters in a string
3. `ord()` - Finds the Unicode number of a character
4. `chr()` - Converts a Unicode number into a character
5. `ascii()` - Shows a safe, readable version of a value

We will explain each one separately, with a simple definition, syntax, and example.


## Section 3: str()

- `str()` converts a value (like a number or boolean) into a string.

**Syntax:**
```python
str(value)
```

````python
# Example 1: Integer to string

a = 25
b = str(a)

# str() has converted the number 25 into the text "25"
print(a)
print(type(b))  # type() tells us the data type
````

````python
# Example 2: Float to string

price = 99.5
price_text = str(price)

# Now price_text holds the string "99.5" instead of the number 99.5
print(price_text)
print(type(price_text))
````

````python
# Example 3: Boolean to string

y = True
z = str(y)

# str() converts True into the text "True"
print(y)
print(type(z))
````

- In all three examples, `str()` took a value that was not text
and turned it into text (a string). This is useful when we want to join numbers with
words, for example while printing messages.


## Section 4: len()

- `len()` returns the number of characters in a string (or items in a list, etc.).

**Syntax:**
```python
len(value)
```


````python
# Example 1: Length of a word

word = "Python"
length_of_word = len(word)

# len() counts every character, including letters
print(length_of_word)  # Output: 6
````

````python
# Example 2: Length of a sentence

sentence = "I love coding"
length_of_sentence = len(sentence)

# len() also counts spaces as characters
print(length_of_sentence)
````

````python
# Example 3: Length of an empty string

empty_text = ""
print(len(empty_text))  # Output: 0, because there are no characters
````

- len()` just tells us "how many characters" are inside a string.
It does not tell us what those characters are or where they are placed (that would be indexing,
which we are not covering in this notebook).


## Section 5: ASCII Value

**What is ASCII?**
- ASCII stands for **American Standard Code for Information Interchange**. It is a system that
gives every character (like letters, digits, and symbols) a unique number.

**What is an ASCII value?**
- An ASCII value is simply the number assigned to a character.

**Why do characters have numeric values?**
- Computers understand numbers, not letters. So every character we type is stored inside the
computer as a number.

- We can find these numbers in Python using the built-in function `ord()`.


````python
# Simple examples of ASCII values using ord()

print(ord("A"))  # ASCII value of uppercase A
print(ord("a"))  # ASCII value of lowercase a
print(ord("0"))  # ASCII value of the digit 0
````

- Notice that `"A"` and `"a"` give different numbers.
- This shows that uppercase and lowercase letters have **different** ASCII values,
- even though they look like the "same" letter to us.


## Section 6: ord()

- `ord()` returns the Unicode code point (number) of a single character.

**Syntax:**
```python
ord(character)
```

````python
# Example: Using ord() on different characters

print(ord("A"))  # Uppercase letter
print(ord("B"))  # Next uppercase letter
print(ord("a"))  # Lowercase letter
print(ord("1"))  # A digit character
````

- `ord()` only works on a single character at a time.
It gives us back a number that represents that character inside the computer.


## Section 7: chr()

- `chr()` converts a Unicode code point (number) into its matching character.

**Syntax:**
```python
chr(number)
```

`chr()` does the opposite job of `ord()`.


````python
# Example: Using chr() to get characters from numbers

print(chr(65))  # Should give us "A"
print(chr(66))  # Should give us "B"
print(chr(97))  # Should give us "a"
````

````python
# Simple relationship between ord() and chr()

print(ord("A"))  # Converts character "A" to number 65
print(chr(65))   # Converts number 65 back to character "A"
````

- `ord()` and `chr()` are opposite functions.
- `ord()` takes a character and gives a number.
- `chr()` takes a number and gives back a character.


## Section 8: ASCII vs Unicode

**Simple explanation:**
- **ASCII** is an older character encoding standard. It only supports a limited number of characters,
  mainly English letters, digits, and a few symbols.
- **Unicode** is a much bigger system. It supports characters from almost every language in the world,
  along with symbols and emojis.
- In Python, `ord()` and `chr()` actually work with **Unicode** code points, not just old ASCII values.
  Since English letters and digits have the same values in both systems, our earlier examples work
  the same way in ASCII and Unicode.

Let's see a simple example with a character outside the basic ASCII range.


````python
# A character outside basic ASCII (a simple Unicode example)

symbol = "€"  # This is the Euro currency symbol
print(symbol)
print(ord(symbol))  # Unicode code point of the Euro symbol
````

**Simple explanation:** The Euro symbol is not part of the old ASCII system, but Python can
still handle it easily because Python uses Unicode.


## Section 9: ascii()

- `ascii()` gives us a safe, readable text version of a value.
-  If the value has characters outside basic ASCII,
-  it shows them using a special escape format instead of the actual character.

**Syntax:**
```python
ascii(value)
```


````python
# Example 1: Normal English text

text1 = "Hello"
print(ascii(text1))  # Normal ASCII text looks unchanged
````

````python
# Example 2: A character outside basic ASCII

text2 = "Café"  # This word contains a special character
print(text2)          # Normal printing shows the actual character
print(ascii(text2))   # ascii() shows a safe, escaped version instead
````

---

**Simple explanation:** `ascii()` is helpful when we want to see a version of our text
that only uses basic ASCII characters, even if the original text has special characters.


## Section 10: Escape Characters

- An escape character (or escape sequence) is a special combination that
starts with a backslash `\` and represents a character that is hard to type directly, like a
new line or a tab space.

### `\n` - New line
- it moves the text to the next line.


````python
# Example of \n (new line)

print("Hello\nPython")
# The \n creates a line break between "Hello" and "Python"
````

### `\t` - Tab
A simple definition: it adds a tab space (like pressing the Tab key).


````python
# Example of \t (tab)

print("Name:\tJohn")
# The \t adds some extra space between "Name:" and "John"
````

### `\\` - Backslash
A simple definition: it is used to print an actual backslash character.


````python
# Example of \\ (backslash)

print("Path: C:\\Python")
# The double backslash prints a single backslash in the output
````

### `\'` - Single quote
A simple definition: it is used to include a single quote inside a single-quoted string.


````python
# Example of \' (single quote)

print('It\'s a sunny day')
# The \' allows us to use an apostrophe inside a single-quoted string
````

### `\"` - Double quote
A simple definition: it is used to include a double quote inside a double-quoted string.


````python
# Example of \" (double quote)

print("She said, \"Python is fun\"")
# The \" allows us to use double quotes inside a double-quoted string
````

### `\b` - Backspace
A simple definition: it removes (moves back over) the character before it.


````python
# Example of \b (backspace)

print("Hello\b World")
# The \b removes the character right before it, shifting the text back
````

### `\r` - Carriage return
A simple definition: it moves the cursor back to the beginning of the current line.


````python
# Example of \r (carriage return)

print("Hello\rWorld")
# The \r moves the cursor to the start of the line, so "World" overwrites "Hello"
````

## Section 11: Raw Strings

**Simple definition:** A raw string tells Python to treat backslashes as normal characters,
instead of treating them as the start of an escape sequence. We create a raw string by adding
an `r` before the quotes.

**Example:**
```python
text = r"Hello\nPython"
```

Let's compare a normal string and a raw string.


````python
# Normal string vs Raw string

normal_text = "Hello\nPython"   # \n here creates a new line
raw_text = r"Hello\nPython"     # r before quotes makes \n a plain text, not a new line

print("Normal string output:")
print(normal_text)

print("\nRaw string output:")
print(raw_text)
````

**Simple explanation:** In the normal string, `\n` created a new line. In the raw string,
`\n` was printed exactly as it is, as plain text, because raw strings ignore escape sequences.
