# Python Math Functions and the Math Module 🧮

### round() Function

- The round() function is used to round a number to the nearest value.

```python
# Example 1:
number = 10.4
print(round(number))   # 10, because 0.4 is less than 0.5
```

```python
number = 10.6
print(round(number))   # 11, because 0.6 is greater than 0.5
```

```python
number = 10.5
print(round(number))   # 10, Python rounds .5 to the nearest even number
```

```python
number = 11.5
print(round(number))   # 12, Python rounds .5 to the nearest even number
```

### Rounding to Decimal Places:

```python
number = 10.5678
print(round(number, 2))   # 10.57, keeps 2 digits after the decimal
```

```python
number = 25.4321
print(round(number, 1))   # 25.4, keeps 1 digit after the decimal
```

```python
percentage = 87.4567
print(round(percentage, 2))   # 87.46, rounds the percentage to 2 decimal places
```

Important:
- round(number) → rounds to the nearest whole number
- round(number, 2) → keeps 2 digits after the decimal
- When the decimal is exactly .5, Python uses "round half to even".

## 2. `abs()`

- Gives the positive value of a number (removes the negative sign).

```python
# Example 1: Absolute value of a negative number
print(abs(-15))   # 15

# Example 2: Absolute value of a positive number stays the same
print(abs(15))    # 15
```

## 3. `pow()`

- Raises a number to a power (x to the power y).

```python
# Example 1: 2 raised to the power 4
print(pow(2, 4))   # 16 (2*2*2*2)

# Example 2: 10 raised to the power 3
print(pow(10, 3))  # 1000
```

## 4. `min()`

- Finds the smallest value from a list of numbers.

```python
# Example 1: Minimum of a few numbers
print(min(10, 4, 8, 2))   # 2

# Example 2: Minimum from a list
marks = [56, 78, 45, 90, 23]
print(min(marks))   # 23
```

## 5. `max()`

- Finds the largest value from a list of numbers.

```python
# Example 1: Maximum of a few numbers
print(max(10, 4, 8, 2))   # 10

# Example 2: Maximum from a list
marks = [56, 78, 45, 90, 23]
print(max(marks))   # 90
```

## 6. `sum()`

- Adds up all the numbers in a list.

```python
# Example 1: Sum of a list of numbers
numbers = [1, 2, 3, 4, 5]
print(sum(numbers))   # 15

# Example 2: Sum of sales data
sales = [200, 450, 300]
print(sum(sales))   # 950
```

# Python `math` Module

- The `math` module is a **built-in module** that gives us extra, ready-made functions for mathematical tasks.

- We must import it before using it.

```python
# Importing the math module
import math

print("math module imported successfully!")
```

## 1. `math.ceil()`

- Rounds a number **UP** to the nearest whole number.

```python
# Example 1: Rounding up a bill
print(math.ceil(45.20))   # 46

# Example 2: Rounding up a negative number
print(math.ceil(-7.9))    # -7
```

## 2. `math.floor()`

- Rounds a number **DOWN** to the nearest whole number.

```python
# Example 1: Rounding down a price
print(math.floor(99.95))   # 99

# Example 2: Rounding down a negative number
print(math.floor(-7.1))    # -8
```

## 3. `math.sqrt()`

- Finds the square root of a number.

```python
# Example 1: Square root of 25
print(math.sqrt(25))   # 5.0

# Example 2: Square root of 2
print(math.sqrt(2))    # 1.41...
```

## 4. `math.pow()`

- Raises a number to a power. Always gives the answer as a float (decimal).

```python
# Example 1: 2 raised to the power 3
print(math.pow(2, 3))   # 8.0

# Example 2: 5 raised to the power 2
print(math.pow(5, 2))   # 25.0
```

## 5. `math.fabs()`

- Gives the positive (absolute) value of a number, always as a float.

```python
# Example 1: Absolute value of a negative number
print(math.fabs(-10))   # 10.0

# Example 2: Absolute value of a temperature drop
temp_change = -3.5
print(math.fabs(temp_change))   # 3.5
```

## 6. `math.factorial()`

- Multiplies a number by every whole number below it, down to 1. (4! = 4×3×2×1)

```python
# Example 1: Factorial of 5
print(math.factorial(5))   # 120

# Example 2: Factorial of 0 (always 1)
print(math.factorial(0))   # 1
```

## 7. `math.gcd()`

- Finds the Greatest Common Divisor — the biggest number that divides two numbers exactly.

```python
# Example 1: GCD of 12 and 18
print(math.gcd(12, 18))   # 6

# Example 2: GCD of 3 numbers
print(math.gcd(24, 36, 60))   # 12
```

## 8. `math.lcm()`

- Finds the Least Common Multiple — the smallest number that is a multiple of two numbers.

```python
# Example 1: LCM of 4 and 6
print(math.lcm(4, 6))   # 12

# Example 2: LCM of 3 numbers
print(math.lcm(3, 4, 5))   # 60
```

## 10. `math.isqrt()`

- Finds the integer (whole number) square root — square root rounded down to a whole number.

```python
# Example 1: Integer square root of 10
print(math.isqrt(10))   # 3

# Example 2: Integer square root of 49
print(math.isqrt(49))   # 7 (exact answer)
```
