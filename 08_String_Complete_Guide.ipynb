## 1️⃣ What is a String?

A **String** is a sequence of characters (letters, numbers, symbols, or spaces) enclosed within quotes.

In Python, the data type used to represent strings is called **`str`**.

### Why do we use Strings?
Strings are used whenever we need to work with **text** — names, sentences, messages, file paths, addresses, emails, and so on.

### What kind of data does a String store?
A String stores **textual data**. Even if it looks like a number (for example `"123"`), as long as it is written inside quotes, Python treats it as text, not as a number.

### Examples of text stored as strings
- `"Lucky Kumar"` → a person's name
- `"Hello, World!"` → a greeting message
- `"123 Main Street"` → an address
- `"2026-09-13"` → a date written as text


````python
# Store text in a variable
name = "Lucky Kumar"

# Print the string
print(name)

# Check its data type
print(type(name))
````

### 🔍 Explaining the Output
- `print(name)` displays the text stored in the variable: `Lucky Kumar`
- `print(type(name))` shows `<class 'str'>`, which confirms that the variable `name` is holding a **String** data type.

This is the foundation of everything else in this notebook — once you understand that a string is simply "text inside quotes," the rest becomes much easier.


## 2️⃣ Creating Strings

Python gives you **four ways** to create a string, depending on what kind of text you need to store.

### a) Single Quotes `' '`
Used for short, simple text.
```python
name = 'Lucky'
```

### b) Double Quotes `" "`
Works exactly the same as single quotes. Useful when your text itself contains a single quote (like `don't`).
```python
name = "Lucky"
```

### c) Triple Single Quotes `''' '''`
Used for text that spans **multiple lines**, or when the text contains both single and double quotes.
```python
message = '''Hello World'''
```

### d) Triple Double Quotes ``
Works the same as triple single quotes. This style is also commonly used for **docstrings** (documentation inside functions/classes).
```python
message = """Hello World"""
```

### When should you use Triple Quotes?
Use triple quotes (`'''...'''` or `"""..."""`) when:
- Your text needs to span **multiple lines**
- Your text contains **both** `'` and `"` characters
- You are writing a **docstring** to describe what a function or program does

For a normal single-line piece of text, single or double quotes are enough.


````python
# a) Single quotes -> good for short, simple text
name1 = 'Lucky'
print(name1)

# b) Double quotes -> works the same as single quotes
name2 = "Lucky"
print(name2)

# c) Triple single quotes -> useful for multi-line text
message1 = '''Hello World'''
print(message1)

# d) Triple double quotes -> same purpose as triple single quotes
message2 = """Hello World"""
print(message2)

# Both single and double quotes create the exact same 'str' type
print(type(name1), type(name2), type(message1), type(message2))
````

### 🔍 Why does the text contain a single quote example matter?

If your sentence contains an apostrophe like `don't`, using single quotes would cause a problem, because Python would think the string ends right after `don`. Double quotes solve this easily.


````python
# This works fine because double quotes are used
sentence = "I don't like rainy days"
print(sentence)

# This would cause an ERROR if uncommented, because Python gets confused
# about where the string ends:
# sentence = 'I don't like rainy days'
````

## 3️⃣ Multiline Strings

When you need to store text that spans **several lines** (like a paragraph, a poem, or a message), triple quotes are the best choice.


````python
# A multiline string using triple double quotes
paragraph = """Dear Team,
This is a reminder about tomorrow's meeting.
Please be on time.
Thank you."""

print(paragraph)
````

Notice how the line breaks you typed inside the triple quotes are preserved exactly as written when the string is printed. This makes triple quotes perfect for emails, messages, or any multi-line text block.


## 4️⃣ Escape Sequences

Sometimes you want to include special characters inside a string — like a new line, a tab space, or a quote character itself. These are written using a **backslash `\`** followed by a special character, and are called **escape sequences**.

| Escape Sequence | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab space |
| `\\` | A single backslash |
| `\'` | A single quote |
| `\"` | A double quote |


````python
# \n -> moves the text to a new line
print("Hello\nWorld")

# \t -> adds a tab space
print("Name:\tLucky")

# \' -> allows a single quote inside a single-quoted string
print('It\'s a sunny day')

# \" -> allows a double quote inside a double-quoted string
print("She said, \"Good morning\"")

# \\ -> prints an actual backslash
print("File path: C:\\Users\\Lucky")
````

## 5️⃣ Raw Strings

Sometimes you want Python to treat backslashes as **plain, literal characters** instead of the start of an escape sequence. This is very useful for things like file paths or regular expressions.

You create a raw string by placing an **`r`** right before the opening quote.


````python
# Normal string -> \n is treated as a new line
normal_path = "C:\new_folder\test"
print("Normal string output:")
print(normal_path)

print()

# Raw string -> backslashes are treated as plain characters
raw_path = r"C:\new_folder\test"
print("Raw string output:")
print(raw_path)
````

Notice how in the normal string, `\n` was accidentally treated as a new line, but in the raw string, everything is printed exactly as typed. This is why raw strings are commonly used for **file paths** on Windows.


## 6️⃣ String Concatenation

**Concatenation** simply means **joining two or more strings together**, usually using the `+` operator.


````python
first_name = "Lucky"
last_name = "Kumar"

# Joining two strings using +
full_name = first_name + " " + last_name
print(full_name)

# Concatenation also works with more than two strings
greeting = "Hello" + ", " + "welcome" + " " + "back!"
print(greeting)
````

⚠️ **Important:** You cannot directly concatenate a string with a number using `+`. Python will raise an error because it does not automatically convert data types during concatenation.


````python
age = 21

# This will cause an ERROR because you are mixing str and int with +:
# message = "My age is " + age

# To fix it, the number must first be converted into a string
message = "My age is " + str(age)
print(message)

# Note: str() is a String Function — it will be covered in detail
# in a separate notebook. Here it is only used to fix the error above.
````

## 7️⃣ String Repetition

The `*` operator can be used to **repeat a string multiple times**.


````python
# Repeating a string using *
laugh = "ha" * 3
print(laugh)

# Repeating a symbol to create a separator line
line = "-" * 20
print(line)

# Repeating a word with spaces
word = "Python " * 4
print(word)
````

## 8️⃣ String Indexing

Every character in a string has a **position number**, called an **index**. Indexing lets you access a single character from a string.

- Indexing starts from **0** (not 1) when counting from the left.
- Indexing starts from **-1** when counting from the right.

```
 P    y    t    h    o    n
 0    1    2    3    4    5      <- positive index (left to right)
-6   -5   -4   -3   -2   -1      <- negative index (right to left)
```


````python
word = "Python"

# Accessing characters using positive index
print(word[0])   # First character
print(word[3])   # Fourth character

# Accessing characters using negative index
print(word[-1])  # Last character
print(word[-2])  # Second-last character
````

⚠️ If you try to access an index that does not exist in the string, Python will raise an `IndexError`.


````python
word = "Python"

# This line is commented out because it would cause an error:
# print(word[10])   # IndexError: string index out of range

# Python has only 6 characters (index 0 to 5), so index 10 does not exist.
print("Valid indexes for 'Python' are from 0 to 5, or -1 to -6")
````

## 9️⃣ String Slicing

**Slicing** lets you extract a **portion (a sub-string)** from a string, instead of just one character.

The general format is:
```
string[start : stop : step]
```
- `start` → index to begin from (included)
- `stop` → index to stop at (**not** included)
- `step` → how many characters to jump each time (optional)


````python
word = "Python Programming"

# Extract characters from index 0 up to (but not including) index 6
print(word[0:6])     # Python

# Extract characters from index 7 to the end
print(word[7:])      # Programming

# Extract from the beginning up to index 6
print(word[:6])      # Python

# Extract the entire string (a full copy)
print(word[:])       # Python Programming

# Using a step value to skip characters
print(word[0:6:2])   # Pto  (every second character)

# Using a negative step to reverse the string
print(word[::-1])    # gnimmargorP nohtyP
````

## 🔒 String Immutability

Strings in Python are **immutable**, which means **once a string is created, it cannot be changed**. You cannot modify individual characters of an existing string.

If you need a "modified" version of a string, Python actually creates a **brand-new string** rather than changing the original one.


````python
word = "Cat"

# Trying to change a character directly is NOT allowed:
# word[0] = "B"   # This raises: TypeError: 'str' object does not support item assignment

# Instead, we build a new string using slicing and concatenation
new_word = "B" + word[1:]
print(new_word)   # Bat

# The original string remains completely unchanged
print(word)       # Cat
````

This behavior is important to understand: every time you seem to "modify" a string in Python, you are really creating a **new string object** in memory. The original one stays exactly as it was.


## 1️⃣1️⃣ Membership Operators (`in` / `not in`)

You can check whether a certain character or a sub-string exists **inside** a string using the `in` and `not in` keywords. These checks return a **Boolean** value: `True` or `False`.


````python
sentence = "I love Python programming"

# Checking if a word exists inside the string
print("Python" in sentence)        # True
print("Java" in sentence)          # False

# Checking that something does NOT exist inside the string
print("Java" not in sentence)      # True
print("love" not in sentence)      # False
````

## 1️⃣2️⃣ Looping Through a String

Since a string is simply a **sequence of characters**, you can loop through it character by character using a `for` loop.


````python
word = "Hi!"

# This loop runs once for every character in the string
for character in word:
    print(character)
````

````python
word = "Code"

# Looping with an index counter using enumerate
for index, character in enumerate(word):
    print("Index", index, "->", character)
````

## 1️⃣3️⃣ Comparing Strings

Strings can be compared using standard comparison operators: `==`, `!=`, `<`, `>`, `<=`, `>=`.

- `==` checks if two strings are exactly equal.
- `<` and `>` compare strings **alphabetically**, based on the underlying character codes.
- Python is **case-sensitive**, meaning `"Apple"` and `"apple"` are considered different strings.


````python
word1 = "apple"
word2 = "Apple"
word3 = "apple"

# Checking exact equality
print(word1 == word3)   # True  (identical text)
print(word1 == word2)   # False (different case: 'a' vs 'A')

# Alphabetical comparison
print("apple" < "banana")   # True, because 'a' comes before 'b'
print("Zebra" < "apple")    # True, because uppercase letters have smaller codes than lowercase
````

## 1️⃣4️⃣ String Formatting (f-strings and `%` operator)

String formatting means **inserting variables or values into a string** in a clean and readable way. Python offers a few approaches; here we cover two of the most common **language features** (not methods).

### a) f-strings (formatted string literals) — the modern, recommended way
Place an `f` right before the opening quote, and put variables inside curly braces `{}`.

### b) `%` operator — the older, traditional way
Uses `%s` for strings, `%d` for integers, and `%f` for decimal numbers.


````python
name = "Lucky"
age = 21

# a) Using an f-string
message1 = f"My name is {name} and I am {age} years old."
print(message1)

# f-strings can also do calculations inside the braces
print(f"Next year I will be {age + 1} years old.")

# b) Using the % operator (older style)
message2 = "My name is %s and I am %d years old." % (name, age)
print(message2)
````

f-strings are generally preferred in modern Python because they are shorter, more readable, and easier to work with.


## 1️⃣5️⃣ The `eval()` Function

`eval()` is a special built-in Python function that takes a **string** and **evaluates (executes) it as if it were actual Python code**.

### Syntax
```python
eval(expression_as_string)
```

### Why is `eval()` useful with strings?
Normally, a string like `"5 + 3"` is just plain text — Python will not calculate it automatically. `eval()` allows you to take that text and actually run it as a real Python expression, returning the calculated result.


````python
# A simple string that looks like a math expression
expression = "5 + 3"

# print() alone treats it as plain text
print(expression)          # Output: 5 + 3   (just text, not calculated)

# eval() actually calculates the expression written inside the string
result = eval(expression)
print(result)               # Output: 8       (an actual number, calculated!)

# Checking the difference in data types
print(type(expression))     # <class 'str'>  -> still text
print(type(result))         # <class 'int'>  -> a real, calculated number
````

````python
# eval() can evaluate more complex expressions too
print(eval("10 * 2 + 5"))      # 25
print(eval("(4 + 6) / 2"))     # 5.0
print(eval("2 ** 3"))          # 8  (2 raised to the power 3)

# eval() can also use variables that already exist in your program
x = 10
y = 5
print(eval("x + y"))           # 15, because it uses the existing x and y
````

````python
# eval() can even evaluate other Python data types written as strings
list_text = "[1, 2, 3, 4]"
actual_list = eval(list_text)

print(actual_list)          # [1, 2, 3, 4]
print(type(actual_list))    # <class 'list'>  -> it became a REAL list, not text!
````

### ⚠️ A Word of Caution About `eval()`

`eval()` is powerful, but it comes with an important warning:

- `eval()` will run **any** valid Python code passed to it — including code that could be harmful, if it comes from an untrusted source (like user input from the internet).
- For this reason, `eval()` should **never** be used on input that you do not fully trust.
- `eval()` is best suited for simple, controlled situations — such as building a basic calculator, or evaluating expressions that your own program generates.

**Good beginner rule:** only use `eval()` on strings that you have written yourself or that come from a fully trusted and controlled source.


````python
# A simple example: a mini calculator using eval()
expression = "12 + 8 * 2"

print("Expression:", expression)
print("Result:", eval(expression))

# eval() correctly follows the normal order of operations (BODMAS/PEMDAS)
# Multiplication happens before addition, so the result is 12 + 16 = 28
````

## 1️⃣6️⃣ Practical Examples

Let's combine everything learned so far into a few small, realistic examples.


````python
# Example 1: Building a formatted greeting message

first_name = "Lucky"
last_name = "Kumar"

# Concatenation to build a full name
full_name = first_name + " " + last_name

# f-string to build a friendly greeting
greeting = f"Hello, {full_name}! Welcome aboard."

print(greeting)
````

````python
# Example 2: Checking and slicing an email address

email = "lucky.kumar@example.com"

# Membership check: does the email contain '@'?
print("@" in email)          # True

# Slicing: extract everything before the '@' symbol
at_position = email.index("@")   # index() is a method, used here only to locate the position
username = email[:at_position]
print("Username part:", username)
````

````python
# Example 3: Using eval() to build a tiny calculator

first_number = 15
second_number = 4
operation = "+"     # can be "+", "-", "*", or "/"

# Building the expression as a string
expression = f"{first_number} {operation} {second_number}"
print("Expression:", expression)

# Evaluating the string expression using eval()
answer = eval(expression)
print("Answer:", answer)
````
