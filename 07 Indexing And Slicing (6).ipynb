# 🐍 Python Indexing & Slicing 

> 💡 **Remember:** Python counting always **starts from 0**, not 1!


## 1️⃣ What is Indexing?

**Indexing** means picking out a **single element** from a sequence (like a list or string) using its position number, called an **index**.

 Indexing always starts at **0**, not 1.

 ```
 Value :   P    y    t    h    o    n
 Index :   0    1    2    3    4    5
 Neg   :  -6   -5   -4   -3   -2   -1
 ```

 Let's try it with a list of fruits 👇

````python
# Value                            =>       P    y    t    h    o    n
# Positive Index (Left to Right)   =>       0    1    2    3    4    5
# Negative Index (Right to Left)   =>      -6   -5   -4   -3   -2   -1
````

````python
fruits = ["apple", "banana", "cherry", "mango", "orange"]

print(fruits[0])   # first element -> apple
print(fruits[1])   # second element -> banana
print(fruits[4])   # last element -> orange
````

### 🔻 Negative Indexing

Python also lets you count **from the end** using negative numbers.

```
Value :   P    y    t    h    o    n
Index :   0    1    2    3    4    5
Neg   :  -6   -5   -4   -3   -2   -1
```

- `-1` → the **last** element
- `-2` → the **second-last** element


````python
fruits = ["apple", "banana", "cherry", "mango", "orange"]

print(fruits[-1])   # last element -> orange
print(fruits[-2])   # second-last element -> mango
````

### ✍️ Indexing Works on Strings Too


````python
name = "Python"

print(name[0])    # P
print(name[3])    # h
print(name[-1])   # n (last character)
````

### ⚠️ Watch Out — IndexError

If you use an index that's too big for the sequence, Python will raise an error.


````python
fruits = ["apple", "banana", "cherry"]

# This line would raise an error because there are only 3 elements (index 0, 1, 2)
# print(fruits[5])   # -> IndexError: list index out of range

print(len(fruits))   # use len() to check how many elements are in the list
````

## 2️⃣ What is Slicing?

- **Slicing** means extracting a **part (multiple elements)** from a sequence such as a **string, list, or tuple** using a range of indexes.

- Slicing allows us to select elements from a **start index to a stop index**, with an optional **step**.

### 📌 Syntax
```python
sequence[start : stop : step]
```

- **start** → where to begin (included)
- **stop** → where to end (⚠️ this position is **NOT included**)
- **step** → how many positions to move or jump each time (default is 1)

All three parts are **optional** — you can leave any of them out.


````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]
#  index:    0   1   2   3   4   5   6   7   8

print(numbers[2:5])    # from index 2 up to (not including) index 5 -> [30, 40, 50]
print(numbers[0:3])    # from index 0 up to (not including) index 3 -> [10, 20, 30]
````

### 🔹 Skipping `start` or `stop`

- `list[:5]` → from the **beginning** to index 4 (no start means "start from the beginning")
- `list[3:]` → from index 3 to the **end** (no stop means "go all the way to the end")
- `list[:]`  → a full copy of the list


````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(numbers[:4])     # start to index 3 -> [10, 20, 30, 40]
print(numbers[5:])     # index 5 to the end -> [60, 70, 80, 90]
print(numbers[:])      # everything -> the whole list
````

### 🔹 Using `step`

- **step** → how many positions to move or jump each time (default is 1)

````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(numbers[0:9:2])   # every 2nd element -> [10, 30, 50, 70, 90]
print(numbers[::3])     # start & stop skipped, every 3rd element -> [10, 40, 70]
````

### 🔻 Negative Slicing

You can use negative numbers in slicing too — to count from the end.


````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(numbers[-3:])     # last 3 elements -> [70, 80, 90]
print(numbers[:-3])     # everything except the last 3 -> [10, 20, 30, 40, 50, 60]
print(numbers[::-1])    # the whole list, reversed! -> [90, 80, ..., 10]
````

### ✍️ Slicing Works on Strings Too


````python
text = "Python Programming"

print(text[0:6])     # -> Python
print(text[7:])      # -> Programming
print(text[::-1])    # reverse the string -> "gnimmargorP nohtyP"
````

## 3️⃣ Indexing vs Slicing — Quick Comparison

| | Indexing | Slicing |
|---|---|---|
| What it does | Selects a single element | Extracts multiple elements |
| Syntax | `x[i]` | `x[start:stop:step]` |
| Example | `fruits[0]` → `"apple"` | `fruits[0:2]` → `["apple", "banana"]` |
| Result | Single element | A sequence of elements |
| Out of range | Raises `IndexError` | Does not raise `IndexError` |

### 💡 Remember

**Indexing → One element**

**Slicing → Multiple elements**

---
## 📝 Summary

- **Indexing** grabs one element (`x[i]`) — counting starts at 0.
- **Negative indexing** counts from the end — `-1` is the last element.
- **Slicing** grabs a chunk (`x[start:stop:step]`) — the `stop` position is always excluded.
- `step` lets you skip elements; a negative `step` reverses the sequence.
- These rules work the same on **strings, lists, and tuples**.
