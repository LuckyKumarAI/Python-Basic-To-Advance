# Introduction to Lists

### What is a List?
- A List is a collection used to store multiple values in a single variable.
- Lists are ordered.
- Lists are mutable, meaning we can change their values.
- Lists allow duplicate values.
- Lists are written using square brackets [ ].
- Can store different data types (Heterogeneous).

### List syntax
A list is written using **square brackets `[]`**, with values separated by commas.
```python
my_list = [value1, value2, value3]
```

````python
# Creating an empty list
empty_list = []
print(empty_list)   # []
````

````python
# Creating a list with values
fruits = ["apple", "banana", "mango"]
print(fruits)   # ['apple', 'banana', 'mango']
````

### Different data types inside a list
A list can hold different data types together — numbers, text, booleans, even other lists.

````python
# A list with mixed data types
mixed = [10, "hello", 3.14, True]
print(mixed)
````

### Duplicate values
A list can have the same value more than once.

````python
# List with duplicate values
numbers = [1, 2, 2, 3, 3, 3]
print(numbers)   # duplicates are allowed
````

### Ordered nature of Lists
Lists keep the **order** in which items were added. The order does not change on its own.

````python
# Order stays the same as we typed it
colors = ["red", "green", "blue"]
print(colors)   # order is preserved
````

### Mutable nature of Lists
Lists are **mutable**, which means we can change their content after creating them (add, remove, or update items).

````python
# Lists can be changed after creation
numbers = [1, 2, 3]
numbers[0] = 100   # changing the first value
print(numbers)     # [100, 2, 3]
````

# Accessing List Elements

We can access list items using their **index** (position number). Indexing starts from **0**.

### Positive indexing
Counting starts from `0` (left to right).

````python
fruits = ["apple", "banana", "mango", "grapes"]

print(fruits[0])   # apple (first item)
print(fruits[2])   # mango (third item)
````

### Negative indexing
Counting starts from `-1` (right to left). Useful for accessing items from the end.

````python
fruits = ["apple", "banana", "mango", "grapes"]

print(fruits[-1])   # grapes (last item)
print(fruits[-2])   # mango (second last item)
````

### First, last and specific elements

````python
fruits = ["apple", "banana", "mango", "grapes"]

print("First item:", fruits[0])
print("Last item:", fruits[-1])
print("Specific item (index 1):", fruits[1])
````

# 3Changing List Elements

Since lists are mutable, we can change their values directly using index positions.

### Change one element

````python
fruits = ["apple", "banana", "mango"]

fruits[1] = "orange"   # changing banana to orange
print(fruits)   # ['apple', 'orange', 'mango']
````

### Change multiple elements (using slicing)

````python
numbers = [1, 2, 3, 4, 5]

numbers[1:3] = [20, 30]   # replacing index 1 and 2
print(numbers)   # [1, 20, 30, 4, 5]
````

# 4. Built-in Functions with Lists

### `len()`
Gives the number of items in a list.

````python
marks = [80, 90, 70]
print(len(marks))   # 3
````

### `min()`
Gives the smallest value in a list.

````python
marks = [80, 90, 70]
print(min(marks))   # 70
````

### `max()`
Gives the largest value in a list.

````python
marks = [80, 90, 70]
print(max(marks))   # 90
````

### `sum()`
Adds up all the values in a list.

````python
marks = [80, 90, 70]
print(sum(marks))   # 240
````

### `sorted()`
Returns a **new sorted list**, without changing the original list.

````python
marks = [80, 90, 70]
sorted_marks = sorted(marks)

print("Original:", marks)          # [80, 90, 70] -> unchanged
print("Sorted (new list):", sorted_marks)   # [70, 80, 90]
````

### `list()`
Converts another data type (like a string or range) into a list.

````python
text = "abc"
print(list(text))   # ['a', 'b', 'c']

numbers = list(range(1, 5))
print(numbers)   # [1, 2, 3, 4]
````

````python
nums = [3, 1, 2]

sorted_nums = sorted(nums)   # does not touch original
print("Using sorted():", sorted_nums, "| Original:", nums)

nums.sort()   # changes original
print("Using sort():", nums)
````

# 5. Checking Elements in a List

We use `in` and `not in` to check if a value exists in a list.

````python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)       # True
print("grapes" in fruits)      # False
print("grapes" not in fruits)  # True
````

### `if` condition with Lists

````python
fruits = ["apple", "banana", "mango"]

if "banana" in fruits:
    print("Banana is available!")
else:
    print("Banana is not available.")
````

# 6. Nested Lists

### What is a nested list?
A **nested list** is a list that contains other lists inside it (list within a list).

### Creating nested lists

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix)
````

### Accessing elements from nested lists

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print(matrix[0])      # first inner list -> [1, 2, 3]
print(matrix[1][2])   # 2nd inner list, 3rd item -> 6
````

### Nested list indexing
The first `[]` selects the inner list, the second `[]` selects the item inside it.

# 7. List Unpacking

Unpacking means assigning list values directly to separate variables.

### Basic unpacking

````python
fruits = ["apple", "banana", "mango"]

a, b, c = fruits   # each variable gets one value
print(a)   # apple
print(b)   # banana
print(c)   # mango
````

### Multiple variable unpacking (with different data)

````python
student = ["Amit", 20, "Data Analyst"]

name, age, role = student
print(name, age, role)
````

### Using `*` in unpacking
The `*` collects extra values into a list.

````python
numbers = [1, 2, 3, 4, 5]

first, *middle, last = numbers
print("First:", first)    # 1
print("Middle:", middle)  # [2, 3, 4]
print("Last:", last)      # 5
````

# 8. Comparing Lists

We can compare lists using `==` and `!=`. This checks if the **values are the same**, not memory location.

````python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = [1, 2, 4]

print(list1 == list2)   # True (same values, same order)
print(list1 == list3)   # False (different values)
print(list1 != list3)   # True
````

# 9. Combining Lists

### Using `+` operator
Joins two lists and creates a **new list**.

````python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

combined = list1 + list2
print(combined)   # [1, 2, 3, 4, 5, 6]
print(list1)      # [1, 2, 3] -> unchanged
````

# 10. Repeating Lists

The `*` operator repeats a list's items multiple times.

````python
numbers = [1, 2, 3]

repeated = numbers * 3
print(repeated)   # [1, 2, 3, 1, 2, 3, 1, 2, 3]
````

````python
# Another example
zeros = [0] * 5
print(zeros)   # [0, 0, 0, 0, 0]
````
