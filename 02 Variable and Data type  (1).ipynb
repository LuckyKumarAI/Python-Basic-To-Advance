## 1. Data Types
Data types define the kind of value that can be stored and used in Python.

```python
name = "Lucky"      # Text => String (short name: str data type)
```

```python
age = 19            # Number => Integer (short name: int data type)
```

```python
height = 5.8        # Decimal => Float (short name: float data type)
```

```python
is_student = True   # True / False => Boolean (short name: bool data type)
```

### Checking Data Types

```python
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

## 2. Variables
Definition: Variables are used to store values in Python.


```python
a = 10
print(a)
```

```python
b = 10.5
print(b)
```

```python
c = "India"
print(c)
```

```python
print(a + 10)
print(type(a))
```

```python
print(b + 10.5)
print(type(b))
```

```python
print(c + " Is Great")
print(type(c))
```
