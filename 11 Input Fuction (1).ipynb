# Python Input



## 1. What is Input?

`input()` is used to take data from the user while the program is running.

#### **Why we use input:**
- To let the user type something into our program (like their name, age, or city).
- This makes our program interactive instead of using fixed values every time.

#### **How input works:**
- When Python runs the `input()` function, the program stops and waits.
- The user types something on the keyboard and presses Enter.
- Whatever the user types is given back to the program.

#### **What type of value does `input()` return?**
- `input()` always gives back the value as a **string** (text), even if the user types numbers.



````python
# This program asks the user to enter their name
name = input("Enter your name: ")  # input() takes what the user types
print(name)  # print() shows the value on the screen
````

## 2. Basic Input

Let's look at another simple example and understand it step by step.


````python
# Ask the user for their name and print a message with it
name = input("Enter your name: ")  # store the user's answer in "name"
print("Your name is:", name)  # print a message using that value
````

**Step-by-step explanation:**
1. Python shows the message `"Enter your name: "` on the screen.
2. The program waits until the user types something and presses Enter.
3. Whatever the user typed is stored inside the variable `name`.
4. `print()` then shows the message `"Your name is:"` followed by the value of `name`.


## 3. Input Always Returns String

This is a very important rule:

> `input()` always returns the user's input as a **string**, even if the user types numbers.


````python
# Ask the user for their age
age = input("Enter your age: ")  # even if user types 20, it is stored as text

print(age)        # shows the value
print(type(age))  # shows the data type of the value
````

Even if you type `20`, the output of `type(age)` will be:

```python
<class 'str'>
```

This means the value is stored as text (string), not as a number.


## 4. Different Types of Input

`input()` always returns a string. But sometimes we want the value as a number.
To do this, we use **type casting**, which means converting one data type into another.

We will look at three cases:
- String input
- Integer input
- Float input


### String Input

Here we simply take the input as it is (text).


````python
# Taking simple text input
name = input("Enter your name: ")  # stored as a string
print(name)
````

### Integer Input

Here we convert the input into a whole number using `int()`.


````python
# Taking integer input using type casting
age = int(input("Enter your age: "))  # input() gives string, int() converts it to a number
print(age)
print(type(age))  # this will show <class 'int'>
````

### Float Input

Here we convert the input into a decimal number using `float()`.


````python
# Taking float (decimal) input using type casting
salary = float(input("Enter your salary: "))  # input() gives string, float() converts it to decimal number
print(salary)
print(type(salary))  # this will show <class 'float'>
````

## 5. Input + Type Casting

### Why do we use type casting with input?

If we want to use the value as a number (to do maths, comparisons, etc.),
we need to **convert** it from string to a number.
This process is called **type casting**.

Example:

```python
age = int(input("Enter your age: "))
```

**Simple steps of what happens here:**
1. `input()` takes the value from the user.
2. The value initially comes as a string.
3. `int()` converts that string into an integer (whole number).


````python
# Integer type casting example
age = int(input("Enter your age: "))  # convert string input to integer
print(age)
````

````python
# Float type casting example
price = float(input("Enter the price: "))  # convert string input to float (decimal number)
print(price)
````

## 6. Taking Multiple Inputs

We can take more than one input in a program.
Each `input()` line takes one value at a time.


````python
# Taking first name and last name separately
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

print(first_name)
print(last_name)
````

````python
# Taking city and country separately
city = input("Enter your city: ")
country = input("Enter your country: ")

print("City:", city)
print("Country:", country)
````

## 7. Checking Data Types

We can take multiple values with different types together, and then check
their types using `type()`.


### Two integer inputs


````python
# Taking two integer numbers from the user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print(num1)
print(num2)
````

### Two float inputs


````python
# Taking two decimal numbers from the user
price1 = float(input("Enter first price: "))
price2 = float(input("Enter second price: "))

print(price1)
print(price2)
````

### Checking the type of different inputs together


````python
# Taking different types of input and checking their type
name = input("Enter your name: ")             # string
age = int(input("Enter your age: "))          # integer
salary = float(input("Enter your salary: "))  # float

print(type(name))    # <class 'str'>
print(type(age))     # <class 'int'>
print(type(salary))  # <class 'float'>
````

**Explanation:**
- `name` stays as a string, so its type is `str`.
- `age` was converted using `int()`, so its type is `int`.
- `salary` was converted using `float()`, so its type is `float`.


## 8. User-Friendly Input

The message inside `input()` is called a **prompt**.
It helps the user understand what they need to type.

**Bad example (no message):**
```python
name = input()
```
The user will not know what to enter here.

**Better example (with message):**
```python
name = input("Enter your name: ")
```
Here the user clearly knows they should type their name.

Always add a clear message inside `input()` so the program is easy to use.


````python
# Not user-friendly: no message shown to the user
# name = input()

# User-friendly: a clear message is shown
name = input("Enter your name: ")
print(name)
````

## 9. Common Mistakes

Let's look at a few common mistakes beginners make while using `input()`.


### Mistake 1: Forgetting Type Casting

If we do not convert the input, it stays as a string, even if it looks like a number.


````python
# Wrong: age is still a string here
age = input("Enter your age: ")
print(type(age))  # <class 'str'>, even if you typed 20
````

````python
# Correct: age is converted into an integer
age = int(input("Enter your age: "))
print(type(age))  # <class 'int'>
````

**Difference:** In the first example, `age` stays as text (string).
In the second example, `age` becomes a real number (integer) because we used `int()`.


### Mistake 2: Invalid Integer Input

This line expects a number:

```python
age = int(input("Enter your age: "))
```

If the user types something like:

```text
twenty
```

instead of `20`, Python will not be able to convert it into an integer, and this will
cause an error. This happens because `int()` only understands numbers, not words.


### Mistake 3: Invalid Float Input

The same problem can happen with `float()`:

```python
salary = float(input("Enter your salary: "))
```

If the user types text instead of a number (like `"high"`), Python will not be able
to convert it into a decimal number, and this will cause an error.

*Note: We are not fixing these errors here. Handling errors using `try-except`
will be covered in a separate notebook.*
