# Python String Basics
### Concatenation, Repetition & Built-in String Methods

---
This notebook covers the fundamental string operations in Python with definitions and examples:
- String Concatenation
- String Repetition
- `upper()`
- `lower()`
- `title()`
- `swapcase()`
- `capitalize()`
- `len()`

## 1. String Concatenation
**Definition:** Joining two or more strings together using the `+` operator.

```python
first_name = "Steve"
last_name = "Jobs"      
full_name = first_name + " " + last_name
print(full_name)
```

```python
greeting = "Good" + "Morning"
print(greeting)
```

```python
name = "Alex"
age = 25
message = "My name is " + name + " and I am " + str(age) + " years old."        # str() converts a number into a string
print(message)
```

## 2. String Repetition
**Definition:** Repeating a string a given number of times using the `*` operator.

```python
word = "Hi"
print(word * 3)
```

```python
border = "-" * 25
print(border)
print("Welcome!")
print(border)
```

## 3. `upper()`
**Definition:** Converts all letters in a string to uppercase.

```python
text = "hello world"
print(text.upper())
```

```python
email = "john.doe@gmail.com"
print(email.upper())
```

```python
warning = "danger ahead"
print(warning.upper())
```

## 4. `lower()`
**Definition:** Converts all letters in a string to lowercase.

```python
text = "HELLO WORLD"
print(text.lower())
```

```python
city = "New York"
print(city.lower())
```

## 5. `title()`
**Definition:** Capitalizes the first letter of every word in a string.

```python
text = "hello world"
print(text.title())
```

```python
book_name = "the great gatsby"
print(book_name.title())
```

```python
address = "123 main street, new york"
print(address.title())
```

## 6. `swapcase()`
**Definition:** Swaps the case of every letter in a string — uppercase becomes lowercase and lowercase becomes uppercase.

```python
text = "Hello World"                    
print(text.swapcase())
```

```python
text = "PyThOn Is FUN"
print(text.swapcase())
```

```python
password = "MyPass123"
print(password.swapcase())
```

## 7. `capitalize()`
**Definition:** Capitalizes only the first letter of the string and makes the rest lowercase.

```python
text = "hello world"
print(text.capitalize())
```

```python
sentence = "PYTHON IS FUN"
print(sentence.capitalize())
```

```python
name = "rahul sharma"
print(name.capitalize())
```

## 8. `len()`
**Definition:** Returns the total number of characters in a string.

```python
# Example 1: Space is also counted
text = "Hello World"

print(len(text))

# Output: 11
# "Hello" = 5 characters
# Space = 1 character
# "World" = 5 characters
# Total = 5 + 1 + 5 = 11
```

```python
text = "Welcome Python"
print(len(text))                  # len() is used to count the number of characters in a string.
                                    # Spaces are also counted.
```

```python
# Example 3: Two words with two spaces
text = "Hello  World"

print(len(text))

# Output: 12
# "Hello" = 5
# Two spaces = 2
# "World" = 5
# Total = 5 + 2 + 5 = 12
```

```python
password = "Secure@123"
print(len(password))
```

```python
name = "Lucky"

print(len(name))
```

```python
message = "Hello World"
                                    # Spaces are also counted.
print(len(message))
```

```python
# len() counts every character, including spaces

text = "A B C"

print(len(text))

# Output: 5
# A = 1
# Space = 1
# B = 1
# Space = 1
# C = 1
# Total = 5
```

```python
# Spaces are also counted as characters

text = "Hello World"

print(len(text))

# Output: 11
# "Hello" = 5 characters
# Space = 1 character
# "World" = 5 characters
# Total = 5 + 1 + 5 = 11
```
